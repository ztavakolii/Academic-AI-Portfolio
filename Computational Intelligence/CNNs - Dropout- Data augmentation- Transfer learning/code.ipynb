```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from torchvision.datasets import ImageFolder
from torch.utils.data import DataLoader
from torch.utils.tensorboard import SummaryWriter
import numpy as np
import matplotlib.pyplot as plt
import os
import time
from datetime import datetime

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

def set_seed(seed=86):
  torch.manual_seed(seed)
  torch.cuda.manual_seed(seed)
  torch.cuda.manual_seed_all(seed)
  torch.backends.cudnn.deterministic = True
  torch.backends.cudnn.benchmark = False

# set seed to avoid problems in finding suitable values for hyperparameters
set_seed()


augmentation_transform = transforms.Compose([
transforms.RandomResizedCrop(224),
transforms.RandomApply([transforms.RandomVerticalFlip()], p=0.2),
transforms.RandomApply([transforms.RandomHorizontalFlip()], p=0.2),
#transforms.RandomApply([transforms.ColorJitter(brightness=0.2, contrast=0.2)], p=0.2),
transforms.RandomApply([transforms.RandomAutocontrast()], p=0.2),
transforms.RandomApply([transforms.RandomRotation(30)], p=0.2),
transforms.Resize((224,224)),
transforms.ToTensor(),
transforms.Normalize(mean=[0.48, 0.45, 0.40], std=[0.23, 0.22, 0.22])
])

transform = transforms.Compose([
transforms.Resize((224,224)),
transforms.ToTensor(),
transforms.Normalize(mean=[0.48, 0.45, 0.40], std=[0.23, 0.22, 0.22])
])


# using ImageFolder function for image dataset organized into folders based on classes
# we read train , test , validation data from folders (we do not need split data to the train/dev/test sets)
train_dataset = ImageFolder(root=train_folder, transform=transform)
augmented_dataset= ImageFolder(root=train_folder, transform=augmentation_transform)
train_dataset = torch.utils.data.ConcatDataset([train_dataset, augmented_dataset])
val_dataset = ImageFolder(root=valid_folder, transform=transform)
test_dataset = ImageFolder(root=test_folder, transform=transform)

# class_names = train_dataset.classes
# print(f"Classes: {class_names}")
# print(f"Train size: {len(train_dataset)}, Val size: {len(val_dataset)}, Test size: {len(test_dataset)}")

batch_size = 64
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)


class CNNNetwork(nn.Module):
  def __init__(self, num_classes=4, dropout_rate=0.3):

      super(CNNNetwork, self).__init__()

      self.block1 = nn.Sequential(
          nn.Conv2d(3, 64, kernel_size=3, padding=1),
          nn.BatchNorm2d(64),
          nn.ReLU(),
          nn.MaxPool2d(2, 2),
          nn.Conv2d(64, 128, kernel_size=3, padding=1),
          nn.BatchNorm2d(128),
          nn.ReLU(),
          nn.MaxPool2d(2, 2),
          nn.Conv2d(128, 256, kernel_size=3, padding=1),
          nn.BatchNorm2d(256),
          nn.ReLU(),
          nn.MaxPool2d(2, 2),
          nn.Conv2d(256, 512, kernel_size=3, padding=1),
          nn.BatchNorm2d(512),
          nn.ReLU(),
          nn.MaxPool2d(2, 2)
      )

      self.avgpool = nn.AdaptiveAvgPool2d((7, 7))
      self.block2 = nn.Sequential(
          nn.Linear(512 * 7 * 7, 1024),
          nn.ReLU(),
          nn.Dropout(dropout_rate),
          nn.Linear(1024, 512),
          nn.ReLU(),
          nn.Dropout(dropout_rate),
          nn.Linear(512, num_classes)
      )

  def forward(self, x):
      x = self.block1(x)
      x = self.avgpool(x)
      x = torch.flatten(x, 1)
      x = self.block2(x)
      return x


def train_model(model, train_loader, val_loader, test_loader, epochs=100, lr=0.001, tensorboard=True):
  model = model.to(device)

# for TensorBoard
  writer = None
  if tensorboard:
      timestamp = datetime.now().strftime('%Y%m%d\_%H%M%S')
      log_dir = f"runs/cancer_classifier\_{timestamp}"
      writer = SummaryWriter(log_dir)

  criterion = nn.CrossEntropyLoss()
  optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=1e-4)
  scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='max', factor=0.5, patience=3, verbose=True)

  best_val_acc = 0.0
  start_time = time.time()

  for epoch in range(epochs):

      model.train()
      running_loss = 0.0
      correct_num = 0
      total_num = 0

      for inputs, labels in train_loader:
          inputs, labels = inputs.to(device), labels.to(device)

          optimizer.zero_grad()
          predictions = model(inputs)
          loss = criterion(predictions, labels)
          loss.backward()
          optimizer.step()

          running_loss += loss.item()
          _, predicted = predictions.max(1)
          total_num += labels.size(0)
          correct_num += predicted.eq(labels).sum().item()

      train_loss = running_loss / len(train_loader)
      train_acc = 100 * correct_num / total_num


      model.eval()
      val_loss = 0.0
      correct_num2 = 0
      total_num2 = 0

      with torch.no_grad():
          for inputs, labels in val_loader:
              inputs, labels = inputs.to(device), labels.to(device)
              predictions = model(inputs)
              loss = criterion(predictions, labels)

              val_loss += loss.item()
              _, predicted = predictions.max(1)
              total_num2 += labels.size(0)
              correct_num2 += predicted.eq(labels).sum().item()

      val_loss /= len(val_loader)
      val_acc = 100 * correct_num2 / total_num2
      scheduler.step(val_acc)

      # saving best model
      if val_acc > best_val_acc:
          best_val_acc = val_acc
          torch.save(model.state_dict(), 'best.pth')

      print(f'Epoch [{epoch+1}/{epochs}] | '
            f'Train Loss: {train_loss:.4f}, Accuracy: {train_acc:.2f}% | '
            f'Val Loss: {val_loss:.4f}, Accuracy: {val_acc:.2f}%')


      #TensorBoard
      if writer:
          writer.add_scalar('Loss/train', train_loss, epoch)
          writer.add_scalar('Accuracy/train', train_acc, epoch)
          writer.add_scalar('Loss/val', val_loss, epoch)
          writer.add_scalar('Accuracy/val', val_acc, epoch)
          writer.add_scalar('Learning Rate', optimizer.param_groups[0]['lr'], epoch)

  # test the best model saved
  model.load_state_dict(torch.load('best.pth'))
  model.eval()
  correct_num3 = 0
  total_num3 = 0
  test_loss = 0.0
  all_preds = []
  all_labels = []

  with torch.no_grad():
      for inputs, labels in test_loader:
          inputs, labels = inputs.to(device), labels.to(device)
          predictions = model(inputs)
          loss = criterion(predictions, labels)

          test_loss += loss.item()
          _, predicted = predictions.max(1)
          total_num3 += labels.size(0)
          correct_num3 += predicted.eq(labels).sum().item()

          all_preds.extend(predicted.cpu().numpy())
          all_labels.extend(labels.cpu().numpy())

  test_loss /= len(test_loader)
  test_acc = 100 * correct_num3 / total_num3

  print(f'Test Results | Loss: {test_loss:.4f}, Accuracy: {test_acc:.2f}%')

  # TensorBoard
  if writer:
      writer.add_scalar('Loss/test', test_loss, epochs)
      writer.add_scalar('Accuracy/test', test_acc, epochs)
      writer.close()

  plot_metrics(writer.log_dir)
  return model



def plot_metrics(log_dir):
  from tensorboard.backend.event_processing import event_accumulator
  ea = event_accumulator.EventAccumulator(log_dir)
  ea.Reload()

  train_acc = ea.Scalars('Accuracy/train')
  val_acc = ea.Scalars('Accuracy/val')
  train_loss = ea.Scalars('Loss/train')
  val_loss = ea.Scalars('Loss/val')

  plt.figure(figsize=(12, 5))
  plt.subplot(1, 2, 1)
  plt.plot([x.step for x in train_acc], [x.value for x in train_acc], label='Train')
  plt.plot([x.step for x in val_acc], [x.value for x in val_acc], label='Validation')
  plt.title('Accuracy Evolution')
  plt.xlabel('Epochs')
  plt.ylabel('Accuracy (%)')
  plt.legend()


  plt.subplot(1, 2, 2)
  plt.plot([x.step for x in train_loss], [x.value for x in train_loss], label='Train')
  plt.plot([x.step for x in val_loss], [x.value for x in val_loss], label='Validation')
  plt.title('Loss Evolution')
  plt.xlabel('Epochs')
  plt.ylabel('Loss')
  plt.legend()

  plt.tight_layout()
  plt.show()



def train_with_transfer_learning(model, train_loader, val_loader, test_loader, epochs=50, lr=0.0001):

  model = model.to(device)

  for param in model.parameters():
      param.requires_grad = False

  if hasattr(model, 'fc'):
      for param in model.fc.parameters():
          param.requires_grad = True
  elif hasattr(model, 'classifier'):
      for param in model.classifier.parameters():
          param.requires_grad = True
  criterion=nn.CrossEntropyLoss()
  optimizer = optim.Adam(filter(lambda p: p.requires_grad, model.parameters()), lr=lr)
  scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=5, gamma=0.1)


  for epoch in range(epochs):
      model.train()
      for inputs, labels in train_loader:
          inputs, labels = inputs.to(device), labels.to(device)
          optimizer.zero_grad()
          predictions = model(inputs)
          loss = criterion(predictions, labels)
          loss.backward()
          optimizer.step()

      scheduler.step()

      model.eval()
      correct = 0
      total = 0
      with torch.no_grad():
          for inputs, labels in val_loader:
              inputs, labels = inputs.to(device), labels.to(device)
              predictions = model(inputs)
              _, predicted = predictions.max(1)
              total += labels.size(0)
              correct += predicted.eq(labels).sum().item()

      acc = 100 * correct / total

      print(f'Epoch [{epoch+1}/{epochs}] | Val Accuracy: {acc:.2f}%')


  correct_num = 0
  total_num = 0
  with torch.no_grad():
      for inputs, labels in test_loader:
          inputs, labels = inputs.to(device), labels.to(device)
          predictions = model(inputs)
          _, predicted = predictions.max(1)
          total_num += labels.size(0)
          correct_num += predicted.eq(labels).sum().item()

  test_acc = 100 * correct_num / total_num

  print(f'Transfer Learning Test Accuracy: {test_acc:.2f}%')


def main():

  print("Training custom CNN model:")
  model = CNNNetwork(num_classes=4, dropout_rate=0.3)
  trained_model = train_model(
      model,
      train_loader,
      val_loader,
      test_loader,
      epochs=50,
      lr=0.001,
      tensorboard=True
  )


  print("\nTransfer Learning with ResNet50:")
  resnet50 = torchvision.models.resnet50(pretrained=True)

  num_ftrs = resnet50.fc.in_features
  resnet50.fc = nn.Linear(num_ftrs, 4)
  train_with_transfer_learning(resnet50, train_loader, val_loader, test_loader)

if __name__ == "__main__":
  main()
```
