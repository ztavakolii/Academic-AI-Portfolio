## import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

# Solving TSP with ACO

## Define the graph (distance matrix)

```python
cities = ["A", "B", "C", "D"]
n = len(cities)

distances = np.array([
    [0, 1, 6, 5],
    [1, 0, 3, 8],
    [6, 3, 0, 4],
    [5, 8, 4, 0]
])
```

### Heuristic information (1 / distance)

```python
heuristic = 1 / (distances + 1e-10)
np.fill_diagonal(heuristic, 0)
pheromone = np.ones_like(distances, dtype=float)
np.fill_diagonal(pheromone, 0)
```

## Set Parameters

```python
n_ants = 4
alpha = 1
beta = 2
ro = 0.1
Q = 100
```

## Defining a function to calculate the tour length

```python
def tour_length(tour):
    return sum(distances[tour[i], tour[(i+1)%n]] for i in range(n))
```

## Construct one tour

```python

def construct_solution(pheromone, heuristic, alpha, beta):
    tour = []
    unvisited = list(range(n))
    current = np.random.choice(unvisited)
    tour.append(current)
    unvisited.remove(current)

    while unvisited:
        # TODO: compute probabilities and choose next city
        probabilities=[]
        for i in unvisited:
            probabilities.append((pheromone[current][i]**alpha)*(heuristic[current][i]**beta))
        probabilities=np.array(probabilities)
        probabilities/=probabilities.sum()

        next_node=np.random.choice(unvisited,p=probabilities)
        tour.append(next_node)
        unvisited.remove(next_node)
        current=next_node

    return tour
```

## Update pheromone matrix

```python
def update_pheromone(pheromone, tours, lengths, Q, ro):
    # TODO: evaporate and deposit pheromone
    pheromone=pheromone*(1-ro)

    for tour , length in zip(tours,lengths):
        x=Q/length
        for i in range(n):
            pheromone[tour[i],tour[(i+1)%n]]+=x
            pheromone[tour[(i+1)%n],tour[i]]+=x
```

## Run one ACO iteration

```python
tours = [construct_solution(pheromone, heuristic, alpha, beta) for _ in range(n_ants)]
lengths = [tour_length(t) for t in tours]
update_pheromone(pheromone, tours, lengths, Q, ro)

best_idx = np.argmin(lengths)
print("Best tour:", [cities[i] for i in tours[best_idx]])
print("Tour length:", lengths[best_idx])
```
